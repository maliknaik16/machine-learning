Let's take the following equation and compute gradients for this:

$$y = x^5+2x^3+4z+8$$

$$\frac{\partial y}{\partial x}=5x^4 + 6x^2$$

$$\frac{\partial y}{\partial z} = 4 $$

```python
import torch
```

```python
x = torch.tensor(2, dtype=torch.float32, requires_grad=True)
x
```

```python
z = torch.tensor(16, dtype=torch.float32)
z
```

```python
# Add gradient tracking for z.
z.requires_grad_()
```

```python
# Define the equiation.
y = x**5 + 2 * x**3 + 4*z + 8
y
```

```python
# Compute the gradients dy/dx and dy/dz
y.backward()
```

```python
# Should output "104.0" because dy/dx = 5x^4+6x^2
print(f"Gradient at x = 2: {x.grad}")

# Should output "4.0" because dy/dz = 4
print(f"Gradient at z = 2: {z.grad}")
```

### Resetting Gradient

Let us take the following equation for w:

$$w = x^5 + x^2 + 3$$
$$\frac{\partial w}{\partial x} = 5x^4+2x$$

```python
w = x**5 + x**2 + 3
w
```

```python
# Compute gradient of w.
w.backward()
```

```python
# Gradient at x should be 84 because dw/dx = 5x^4+2x
print(f"Gradient at x = 2: {x.grad}")
```

```python
# Delete the tensor
del w

# Initialize the quation again.
w = x**5 + x**2 + 3
```

```python
# The reason we see incorrect value is because we need to reset gradient if we are performing any
# operations on the x with multiple variables. It'll add all the values.
x.grad.zero_()

# Compute the gradient of w again on x.
w.backward()

# Get the gradient.
print(f"Gradient at x = 2: {x.grad}")
```

## Disabling Gradient tracking

There are times where we want to stop the gradient tracking for example during the inference so pytorch provides the `torch.no_grad` to stop the gradient tracking.

```python
# Reset gradient
x.grad.zero_()
```

```python
v = 3 * x**2

v.requires_grad
```

```python
# Use no_grad to disable gradient tracking.
with torch.no_grad():
    v = 3 * x**2

v.requires_grad
```

```python
# or another way is to use detach() method
v = 3 * x**2
v.detach_()
v.requires_grad
```
